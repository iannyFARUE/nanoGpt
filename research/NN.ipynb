```python
import torch 
import torch.nn as nn
```

```python
model = nn.Sequential(
    nn.Linear(in_features=3, out_features=2, bias=True),
     nn.ReLU(),
    nn.Linear(in_features=2, out_features=4, bias=True),
     nn.ReLU(),

    nn.Linear(in_features=4, out_features=2, bias=True),
     nn.Softmax(dim=1),
)
```
