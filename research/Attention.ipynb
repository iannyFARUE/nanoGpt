## Simple Attention mechanism without trainable weights

```python
import torch
inputs = torch.tensor(
 [[0.43, 0.15, 0.89], # Your (x^1)
 [0.55, 0.87, 0.66], # journey (x^2)
 [0.57, 0.85, 0.64], # starts (x^3)
 [0.22, 0.58, 0.33], # with (x^4)
 [0.77, 0.25, 0.10], # one (x^5)
 [0.05, 0.80, 0.55]] # step (x^6)
)
```

```python
input_query = inputs[1]
input_query
```

```python
input_1 = inputs[0]
input_1
```

```python
0.5500 * 0.43 + 0.87 * 0.15 + 0.66 * 0.89
```

```python
torch.dot(input_1,input_query)
```

```python
res = 0
for idx, el in enumerate(inputs[0]):
    res += inputs[0,idx] * input_query[idx]
res
```

```python
attn_scores_2 = torch.empty(inputs.shape[0])
for idx, x_i in enumerate(inputs):
    attn_scores_2[idx] = torch.dot(input_query, x_i)
attn_scores_2
```

```python
attn_weights_2_temp = attn_scores_2 / attn_scores_2.sum()
attn_weights_2_temp
```

```python
def softmax_naive(x):
    return torch.exp(x) / torch.exp(x).sum(dim=0)
```

```python
softmax_naive(attn_scores_2)
```

```python
torch.softmax(attn_scores_2,dim=0)
```

```python
attn_weights_2 = torch.nn.functional.softmax(attn_scores_2,dim=0)
```

```python
attn_weights_2
```

```python
query = inputs[1]
context_vec_2 = torch.zeros(query.shape)
for i, x_i in enumerate(inputs):
    context_vec_2 += attn_weights_2[i] * inputs[i]
context_vec_2
```
